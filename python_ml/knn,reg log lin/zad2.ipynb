```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.datasets import make_regression
import seaborn as sns
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

Napisz funkcję compare_errors(y_true, y_pred), która liczy MSE, RMSE, MAE i R², a następnie na sztucznie wygenerowanym zbiorze z jednym mocnym outlierem pokazuje, jak bardzo MAE i MSE reagują inaczej na ten outlier.

```python
X, y = make_regression(n_samples=200, n_features=2, noise=1, random_state=42)
X
```

X – Macierz cech (Features / Predyktory)
Co to jest: Dane wejściowe. Wszystkie czynniki, zmienne i wskaźniki, które dajesz modelowi, żeby na ich podstawie spróbował zrozumieć rynek lub zjawisko.

y – Wektor celu (Target / Zmienna zależna)
Co to jest: Dana wyjściowa. To, co chcesz przewidzieć. Rzeczywisty wynik (Ground Truth), do którego model próbuje się dopasować.

```python
#zmienilismy y bo chcemy zobaczyc jak zachowuja sie bledy z outlinerem
y[0] = 600
y
```

```python
model = LinearRegression()
model.fit(X,y)
```

```python
y_pred = model.predict(X)


def compare_errors(y_true, y_pred):
    
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)


    return f" mse: {mse}, rmse: {rmse}, mae: {mae}, r2: {r2}"

compare_errors(y, y_pred)
```

MAE (Mean Absolute Error – Średni błąd bezwzględny)Co oznacza: Średnia różnica między wartościami przewidywanymi a rzeczywistymi, traktująca każdy błąd z równą wagą.

RMSE (Root Mean Squared Error – Pierwiastek błędu średniokwadratowego)Co oznacza: Odchylenie standardowe reszt (różnic między predykcją a rzeczywistością).

R² (Współczynnik determinacji)Co oznacza: Procent wariancji zmiennej zależnej, który jest wyjaśniany przez model

MSE (Mean Squared Error – Błąd średniokwadratowy) to jedna z najważniejszych i najczęściej stosowanych metryk oceny modeli regresji. Mierzy ona średnią kwadratów błędów, czyli średnią kwadratów różnic między wartościami przewidzianymi przez model (\(\^{y}_{i}\)) a wartościami rzeczywistymi (\(y_{i}\)).
