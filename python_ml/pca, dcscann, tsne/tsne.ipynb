```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
%matplotlib inline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(
    n_samples=500, 
    n_features=3, 
    centers=4, 
    cluster_std=0.6, 
    random_state=42
)
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(projection='3d')

scatter = ax.scatter(X[:, 0], X[:, 1], X[:, 2], 
                     c=y, cmap='viridis', s=50)
```

```python
scaler = StandardScaler()
scaler.fit(X)
scaled_data = scaler.transform(X)
scaled_data.shape
```

```python
pca = PCA(n_components=2)
```

```python
X_reduced = pca.fit_transform(scaled_data)
X_reduced.shape
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(X_reduced[:, 0], X_reduced[:, 1], c=y, cmap='viridis', s=50, alpha=0.8)
plt.grid(True)
```

```python
from sklearn.manifold import TSNE
```

```python
tsne = TSNE(n_components=2,perplexity=40, random_state=42)
X_tsne = tsne.fit_transform(scaled_data)

plt.figure(figsize=(8, 6))
plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis', s=50, alpha=0.8)
plt.grid(True)
plt.show()
```

```python

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

axes[0].scatter(X_reduced[:, 0], X_reduced[:, 1], c=y, cmap='viridis', s=40, alpha=0.8)
axes[0].set_title("PCA (2D)")
axes[0].grid(True)

axes[1].scatter(X_tsne[:, 0], X_tsne[:, 1], c=y, cmap='viridis', s=40, alpha=0.8)
axes[1].set_title("t-SNE (2D)")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
from sklearn.datasets import load_digits
```

```python
digits = load_digits(as_frame=True)
df = digits.frame
df.info()
```

```python
df.describe()
```

```python
X_3 = digits.data
y_3 = digits.target
y_3.value_counts()
```

```python
tsne_3 = TSNE(n_components=2, perplexity=30, random_state=42)
X_tsne_3 = tsne_3.fit_transform(X_3)  

plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_tsne_3[:, 0], X_tsne_3[:, 1], c=y_3, cmap='tab10', s=30, alpha=0.8)

plt.colorbar(scatter, ticks=range(10), label="colors")
plt.title("t-SNE")
plt.grid(True)
plt.show()
```
