```python
from sentence_transformers import SentenceTransformer

# Load a small pretrained model that turns a sentence into a 384-number vector
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

sentences = [
    "The cat sleeps on the sofa.",       # 0: animals
    "A dog is running in the park.",     # 1: animals
    "I love eating pizza.",              # 2: food
    "Pasta is my favourite dinner.",     # 3: food
    "The stock market fell today.",      # 4: finance
]

embeddings = model.encode(sentences)     # shape: (5 sentences, 384 numbers)
print(embeddings.shape)                  # prediction: ?

# Matrix of cosine similarities: row i, column j = similarity of sentence i and j
print(model.similarity(embeddings, embeddings))
```

```python
def search(query, sentences, embeddings, model, k=2):
    """Return the k sentences most similar to the query, with their cosine scores."""
    q_emb = model.encode([query])                          # embed the question (same model!)
    scores = model.similarity(q_emb, embeddings)[0]        # cosine with every sentence
    top = scores.argsort(descending=True)[:k]              # indexes of the k best scores
    return [(sentences[i], scores[i].item()) for i in top]


print(search("What should I cook for dinner?", sentences, embeddings, model))
print(search("what my animal is doing?", sentences, embeddings, model))

# prediction: which two sentences will come first?
```
