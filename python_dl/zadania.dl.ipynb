# zadanie 2 

```python
# y = 2x + 1
# mamy odkryc ze waga to 2 a wyraz wolny to 1 
```

```python
import numpy as np

x = np.array([-5, -2, 0, 1, 2, 3, 4], dtype=float)
y = np.array([-9, -3, 1, 3, 5, 7, 9], dtype=float)
```

```python
model = models.Sequential([
    layers.Input(shape=(1,)),  
    layers.Dense(1)             
])
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.1),
    loss='mse')
```

```python
from keras import optimizers
```

```python
history = model.fit(x,
                    y,
                    epochs = 500,
                    verbose=0)
```

```python
w, b = model.get_weights()
```

```python
w
```

```python
b
```

```python
model.predict(np.array([10.0]))
```

```python
model.predict(np.array([20.0]))
```

```python
model.predict(np.array([-5.0]))
```

```python
loss_function_1 = history.history['loss'][-1]
```

# zadanie 6

```python
tf.keras.utils.set_random_seed(42)
model_5 = models.Sequential([
    layers.Input(shape=(1,)),  
    layers.Dense(16, activation='relu'),
    layers.Dense(1)
])
```

```python
tf.keras.utils.set_random_seed(42)
model_6 = models.Sequential([
    layers.Input(shape=(1,)),  
    layers.Dense(16, activation='relu'),
    layers.Dense(16, activation='relu'),
    layers.Dense(1)
])
```

```python
model_5.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss='mse')
```

```python
model_6.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.01),
    loss='mse')
```

```python
history_5 = model_5.fit(x, y, epochs=500, verbose=0)
```

```python
loss_function_x2_2 = history_5.history['loss'][-1]
```

```python
model_5.predict(np.array([5]))
```

```python
history_6 = model_6.fit(x, y, epochs=500, verbose=0)
```

```python
loss_function_x2_3 = history_6.history['loss'][-1]
```

```python
model_6.predict(np.array([5]))
```
