```python
# y = 2x + 1
# mamy odkryc ze waga to 2 a wyraz wolny to 1 
```

```python
import numpy as np

x = np.array([0, 1, 2, 3, 4], dtype=float)
y = np.array([1, 3, 5, 7, 9], dtype=float)
```

```python
model = models.Sequential([
    layers.Input(shape=(1,)),  
    layers.Dense(1)             
])
```

```python
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.1),
    loss='mse')
```

```python
from keras import optimizers
```

```python
history = model.fit(x,
                    y,
                    epochs = 500,
                    verbose=0)
```

```python
w, b = model.get_weights()
```

```python
w
```

```python
b
```

```python
model.predict(np.array([10.0]))
```

```python
model.predict(np.array([20.0]))
```

```python
model.predict(np.array([-5.0]))
```

```python
history.history['loss'][-1]
```

# zadanie 3 

```python
model_2= models.Sequential([
    layers.Input(shape=(1,)),  
    layers.Dense(4, activation='relu'),
    layers.Dense(1)             
])
```

```python
model_2.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.1),
    loss='mse')
```

```python
history_2 = model_2.fit(x,
                    y,
                    epochs = 500,
                    verbose=0)
```

```python
w1, b1, w2, b2 = model_2.get_weights()
```
